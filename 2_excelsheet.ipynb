```python
#load data
import pandas as pd
file=pd.read_csv("/content/irisexcel.csv")
x=file[["sepal_length","sepal_width","petal_length","petal_width"]]
y=file["species"]
#load model
from sklearn.linear_model import Perceptron
m=Perceptron()
#load data into model
m.fit(x,y)
#testing
op=m.predict([[1.5,3.6,2.9,9.6]])
print(op)
```
